## 1. My lane as an ML task (type)

*Classification, clustering, ranking, or scoring — which one, and why?*

```python
# My lane as an ML task: Structured Content Archetype Clustering
# Task type: Clustering (unsupervised learning)

# I've mostly worked on classification and other supervised learning problems 
# before, and read a lot about clustering, but never built a full clustering 
# project end to end — so this lane is new territory for me.
# Technically, this fits clustering because there's no known outcome or label 
# I'm predicting. I'm not classifying pages or scoring them 
# against a target — I'm trying to discover what natural groups (archetypes) 
# exist in the content, based on how pages perform. That makes it unsupervised, 
# not classification, ranking, or scoring.
```

## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*

```python

# There is no target or label for this task. Clustering is unsupervised I'm 
# not predicting a known outcome, so there's nothing to observe or define a 
# label from.

# Instead of a target, I define the "features" that the clustering will group 
# pages by:
# - visibility (impressions, average position)
# - freshness (days since last update, content age)
# - engagement (engagement rate, scroll rate)
# - content depth (word count)

# The "answer" isn't a predicted label per row — it's the cluster each page 
# lands in, discovered from how these features combine. Only after clustering 
# do I inspect each group and give it a name. 
```

## 3. Success metric

*One metric you can defend. What number means 'good'?*

```python


# One metric for evaluating clustering is the Silhouette score, a value between -1 and 1 measuring how well-separated the clusters are.
# Higher means points are clearly closer to their own cluster than to neighboring ones. 
# I'll treat anything meaningfully above 0 as a reasonable starting point, and compare scores across different numbers of clusters (k) to pick the best split.
# Higher means points are clearly closer to their own cluster than to neighboring ones.
# I'll treat anything meaningfully above 0 as a reasonable starting point, and compare scores across different numbers of clusters (k) to pick the best split.

# Silhouette score alone isn't enough, though a mathematically "clean" split 
# could still be meaningless in practice. So I'll pair it with a human 
# sense-check its pulling a handful of real pages from each cluster and asking 
# whether they actually look alike in a way that matters (e.g. all stale but 
# high-traffic, or all thin content with low engagement). A cluster only counts 
# as "good" if both the number and the manual read-through agree.
```
